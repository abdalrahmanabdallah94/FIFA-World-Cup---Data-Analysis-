```python
#س1 عدد الدول اللي شاركت بفرق رجال ونساء؟
#س2 عدد الدول اللي شارك بفرق رجال بس؟
#س3 عدد الدول اللى شاركت بفرق نساء فقط؟
#س4 مدة كل بطولة؟
#س5 أكثر منتخبات  وصلت للمربع الذهبي

df_tournaments = tournaments.copy()
df_squads = squads.copy()
df_standings = tournament_standings.copy()

for df in [df_squads, df_standings]:
    if 'gender' not in df.columns and 'female' in df.columns:
        df['gender'] = df['female'].map({0: 'Men', 1: 'Women', False: 'Men', True: 'Women'})

#س1 #س2 #س3
men_teams = set(df_squads[df_squads['gender'] == 'Men']['team_name'].unique())
women_teams = set(df_squads[df_squads['gender'] == 'Women']['team_name'].unique())

both_genders_count = len(men_teams.intersection(women_teams))
men_only_count = len(men_teams.difference(women_teams))
women_only_count = len(women_teams.difference(men_teams))

df_result = pd.DataFrame({
    'Category': [
        'Count countries shared both genders',
        'Count countries shared men genders',
        'Count countries shared women genders'
    ],
    'Count': [
        both_genders_count,
        men_only_count,
        women_only_count
    ]
})

display(df_result)

#س4
df_tournaments['start_date'] = pd.to_datetime(df_tournaments['start_date'])
df_tournaments['end_date'] = pd.to_datetime(df_tournaments['end_date'])
df_tournaments['duration_days'] = (df_tournaments['end_date'] - df_tournaments['start_date']).dt.days + 1

top3_men = (
    df_tournaments[df_tournaments['gender'] == 'Men']
    .sort_values(by='duration_days', ascending=False)
    .head(3)[['tournament_name', 'start_date', 'end_date', 'duration_days']]
)

print("Top long time tournaments - Men")
display(top3_men)

top3_women = (
    df_tournaments[df_tournaments['gender'] == 'Women']
    .sort_values(by='duration_days', ascending=False)
    .head(3)[['tournament_name', 'start_date', 'end_date', 'duration_days']]
)

print("Top long time tournaments - Women")
display(top3_women)

##س5
df2 = (
    df_standings[df_standings['position'] <= 4]
    .groupby(['gender', 'team_name'])
    .size()
    .reset_index(name='top4_appearances')
    .sort_values(by=['gender', 'top4_appearances'], ascending=[True, False])
)

top4_men = df2[df2['gender'] == 'Men'].head(3)
top4_women = df2[df2['gender'] == 'Women'].head(3)

print("Most teams in top 4 _ Men")
display(top4_men[['team_name', 'top4_appearances']])

print("Most teams in top 4 _ Women")
display(top4_women[['team_name', 'top4_appearances']])
```
